**Проведем профилирование**

```python
import cProfile
import pstats
import io

with cProfile.Profile() as pr:
    data = run_simulation(
        years=6,
        config_path="config.yaml",
        llm_based_president=False,
        llm_based_households=False,
        president_setup="",
        show_plots=False
    )

s = io.StringIO()
ps = pstats.Stats(pr, stream=s).sort_stats('cumtime')
ps.print_stats('src')

print(s.getvalue())
```

Как видно из профилирования, без llm все работает довольно-таки быстро. Все основные методы вызываются ровно столько раз, сколько и должны

**Проведем профилирование с LLM**

```python
import cProfile
import pstats
import io

with cProfile.Profile() as pr:
    data = run_simulation(
        years=2,
        config_path="config.yaml",
        llm_based_president=True,
        llm_based_households=True,
        president_setup="Ты отличный президент.",
        show_plots=False,
        disable_parse_errors=True
    )

s = io.StringIO()
ps = pstats.Stats(pr, stream=s).sort_stats('cumtime')
ps.print_stats('src')

print(s.getvalue())
```

Вот это да! Профилирование показало, что запросы по потреблению в 3.5 раза более времязатратные. Оказалось что я забыл добавить return в decide_consumption и поэтому каждый раз функция 5 раз вместо одного положенного обращалась к API. Хорошо, что заметил, а то столько денег на этом потерялось бы! Это же целых 4*0.005*years=0.02($/year), то есть по 0,4 бакса на одну симуляцию. Я уже запускал ее раза 4, то есть минус два бакса.
